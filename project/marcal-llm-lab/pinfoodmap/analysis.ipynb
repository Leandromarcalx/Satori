# TikTok Food Data Analysis
This notebook analyzes the extracted food-related videos, including their transcriptions, translations, and identified locations.

```python
import pandas as pd
import os

# File path
csv_file = 'processed_json_enhanced.csv'

if os.path.exists(csv_file):
    df = pd.read_csv(csv_file)
    print(f"Loaded {len(df)} food-related items.")
else:
    print(f"File {csv_file} not found. Please run main.py first.")
```

## Data Overview
Displaying the first few processed videos.

```python
df.loc[df['AI_Category'] == 'restaurant review and recommendation'].head()
```

## Identified Locations & Restaurants
Cleaning up and viewing the extracted entities.

```python
# Filter rows that have identified locations or restaurants
entities_df = df[(df['Locations'].notna() & (df['Locations'] != '')) | 
                 (df['Restaurants'].notna() & (df['Restaurants'] != ''))]

print(f"Found {len(entities_df)} items with specific entities.")
entities_df[['Link', 'Locations', 'Restaurants', 'Translated Content']].head()
```

## Detailed View of Transcriptions
Look at the specific content transcribed from the videos.

```python
for idx, row in df.iterrows():
    print(f"Video {idx+1}: {row['Link']}")
    print(f"Locations: {row['Locations']}")
    print(f"Transcription Snippet: {str(row['Transcription'])[:200]}...")
    print("-" * 50)
```
